# 06 · Fitting with limb darkening integrated out

**Level:** intermediate · **Prerequisites:** [03](03_first_fit.ipynb) · **Run time:** ~3 min (~40 s with `PIZZETTI_SMOKE=1`)

**Covers:** `MarginalTransitModel`, `fit()` and `FitResult` (`summary`, `corner`, weighted draws, diagnostics), `EnsembleSampler`, `expand`, priors (`ld_prior`: physical, flat, Gaussian, callable; `bounds`, `log_prior`), power-2 with free α, secondary eclipses

The flux is *linear* in the limb-darkening coefficients and in the baseline flux. `MarginalTransitModel` uses this to
integrate them out of the likelihood analytically, over their physical region, so that a sampler only explores the geometry
(t0, per, rp, a, b; optionally ecc, w, α). The coefficients are drawn back afterwards for every posterior sample.

We fit the hot Jupiter of notebooks 02 and 03 (7 transits, 2-min cadence, 500 ppm, quadratic limb darkening) three ways:

| | method | what the sampler sees |
|---|---|---|
| **A₀** | example 01 with batman: `batman.TransitModel` + the same `log_prob` + `emcee` (fixed-budget timing only) | t0, per, rp, a, b, q1, q2 (7-D) |
| **A** | the original notebook: `pizzetti.TransitModel` + hand-written `log_prob` + `emcee` | t0, per, rp, a, b, q1, q2 (7-D) |
| **B** | `pizzetti.MarginalTransitModel` + `model.EnsembleSampler` (an `emcee.EnsembleSampler` subclass) | t0, per, rp, a, b (5-D) |
| **C** | `pizzetti.MarginalTransitModel.fit()`: profile importance sampling, no MCMC | t0, per, rp, a, b (5-D) |


In B and C the quadratic coefficients **and** the out-of-transit flux are integrated out over their physical region
(Kipping 2013 conditions, uniform prior on q1, q2). The user code for B differs from A by about three lines.

*Timing note: wall-clock times depend on the machine and its load; evaluation counts and autocorrelation times do not.*

````python
import time
import numpy as np
import matplotlib.pyplot as plt
import emcee
import corner
import batman
import os
import pizzetti

SMOKE = os.environ.get("PIZZETTI_SMOKE") == "1"   # shorter runs in the automatic tests
print("batman", batman.__version__, "| pizzetti", pizzetti.__version__, "| emcee", emcee.__version__)
````

## Data (identical to notebooks 02 and 03)

````python
truth = dict(t0=0.0, per=3.5, rp=0.10, a=9.0, b=0.35, u1=0.40, u2=0.25)
sigma = 500e-6

epochs = np.arange(7)
t_one = np.arange(-0.13, 0.13, 2.0 / 1440)
t = (truth["t0"] + epochs[:, None] * truth["per"] + t_one[None, :]).ravel()

def params(module, t0, per, rp, a, b, u1, u2):
    p = module.TransitParams()
    p.t0, p.per, p.rp, p.a = t0, per, rp, a
    p.inc = np.degrees(np.arccos(b / a))
    p.ecc, p.w = 0.0, 90.0
    p.limb_dark, p.u = "quadratic", [u1, u2]
    return p

p_true = params(pizzetti, **truth)
f_true = pizzetti.TransitModel(p_true, t, series_tol=0.0, limb="quadrature").light_curve(p_true)
y = f_true + sigma * np.random.default_rng(42).standard_normal(t.size)

fold = (t - truth["t0"] + 0.5 * truth["per"]) % truth["per"] - 0.5 * truth["per"]
plt.figure(figsize=(8, 3))
plt.plot(24 * fold, 1e6 * (y - 1), ".", ms=2, color="0.6")
plt.plot(24 * np.sort(fold), 1e6 * (f_true[np.argsort(fold)] - 1), "k", lw=1)
plt.xlabel("hours from mid-transit"); plt.ylabel("flux - 1 (ppm)");
````

## Common settings

Same starting points as notebook 02. All emcee runs continue until the emcee convergence recipe is met
(chain longer than 50 autocorrelation times, τ estimate stable to 1 %), so the comparison is
**effort to convergence**, not effort for a fixed number of steps.

````python
u1, u2 = truth["u1"], truth["u2"]
th0 = np.array([truth["t0"], truth["per"], truth["rp"], truth["a"], truth["b"],
                (u1 + u2)**2, u1 / (2 * (u1 + u2))])
nwalkers = 32
p0 = th0 + np.array([2e-4, 2e-6, 1e-3, 0.05, 0.02, 0.02, 0.02]) * np.random.default_rng(1234).standard_normal((nwalkers, 7))

def run_to_convergence(sampler, p0, ntau=(10 if SMOKE else 50), check=500, max_steps=200_000, seed=2024):
    sampler.random_state = np.random.RandomState(seed).get_state()     # reproducible
    t_start = time.perf_counter()
    old = np.inf
    for _ in sampler.sample(p0, iterations=max_steps):
        if sampler.iteration % check:
            continue
        tau = sampler.get_autocorr_time(tol=0)
        if np.all(ntau * tau < sampler.iteration) and np.all(np.abs(old - tau) / tau < 0.01):
            break
        old = tau
    wall = time.perf_counter() - t_start
    tau = sampler.get_autocorr_time(tol=0)
    return dict(wall=wall, steps=sampler.iteration, evals=sampler.iteration * sampler.nwalkers,
                tau=tau, burn=int(5 * tau.max()))
````

## A. The original notebook: `TransitModel` + hand-written `log_prob` + emcee

````python
names = ["t0", "per", "rp", "a", "b", "q1", "q2"]

def to_u(q1, q2):
    s = np.sqrt(q1)
    return 2 * s * q2, s * (1 - 2 * q2)

prm = params(pizzetti, **truth)
model_A = pizzetti.TransitModel(prm, t)
n_eval_A = 0

def log_prob(th):
    global n_eval_A
    n_eval_A += 1
    t0, per, rp, a, b, q1, q2 = th
    if not (0 < rp < 0.5 and 1.5 < a < 100 and 0 <= b < 1 + rp and 0 < q1 < 1 and 0 < q2 < 1):
        return -np.inf
    prm.t0, prm.per, prm.rp, prm.a = t0, per, rp, a
    prm.inc = np.degrees(np.arccos(b / a))
    prm.u = list(to_u(q1, q2))
    r = y - model_A.light_curve(prm)
    return -0.5 * np.dot(r, r) / sigma**2

log_prob(p0[0])                                       # warm-up (numba compilation)
sampler_A = emcee.EnsembleSampler(nwalkers, 7, log_prob)
res_A = run_to_convergence(sampler_A, p0)
chain_A = sampler_A.get_chain(discard=res_A["burn"], flat=True)
print(f"A: {res_A['steps']} steps, {res_A['evals']} evaluations, max tau {res_A['tau'].max():.0f}, {res_A['wall']:.1f} s")
````

## B. New method, emcee API

`MarginalTransitModel` takes the same `TransitParams` plus the data. Its `EnsembleSampler` is an
`emcee.EnsembleSampler` subclass, so progress bars, backends, `get_autocorr_time` etc. all work as usual.
The old 7-column `p0` is accepted as is (the q1, q2 columns are ignored), and `get_chain` returns
t0, per, rp, a, b **plus** q1, q2 and the baseline f0, drawn from their exact conditional posterior.
Internally the sampler uses the transit duration T14 instead of a/R* (prior kept uniform in a), and all
walkers of a half-ensemble are evaluated in one compiled call.

````python
model = pizzetti.MarginalTransitModel(p_true, t, y, sigma,
                                      free=["t0", "per", "rp", "a", "b"],
                                      bounds={"rp": (0, 0.5), "a": (1.5, 100)})
model.log_prob_internal(model.to_internal(p0[:4, :5]))      # warm-up (numba compilation)
model.n_eval = 0
sampler_B = model.EnsembleSampler(nwalkers)
res_B = run_to_convergence(sampler_B, p0)
res_B["evals"] = model.n_eval
chain_B = sampler_B.get_chain(discard=res_B["burn"], flat=True, seed=1)
print(sampler_B.labels)
print(f"B: {res_B['steps']} steps, {res_B['evals']} evaluations, max tau {res_B['tau'].max():.0f}, {res_B['wall']:.1f} s")
````

## C. New method, one call (no MCMC)

`fit()` finds the posterior maximum, builds a proposal along the impact parameter from conditional optima
(which covers the b ≥ 0 wall and skewed tails), and importance-samples with Pareto-smoothed weights.
If the diagnostics fail (k̂ ≥ 0.7 or an incomplete profile) it falls back to emcee automatically.
`result.samples` are equal-weight draws (systematic resampling); for histograms and corner plots the exact
weighted draws `result.weighted_samples` with `result.weights` are preferable (resampled draws repeat values
in proportion to their weights, which shows up as spikes in fine histograms).

````python
model.n_eval = 0
t_start = time.perf_counter()
result = model.fit(seed=1, n=8000)
wall_C = time.perf_counter() - t_start
print(result.diagnostics)
print(result.summary())
````

## Performance

````python
def ess_per_sec(res):
    # effective samples of the worst-mixing parameter after burn-in, per second of sampling
    n = (res["steps"] - res["burn"]) * nwalkers
    return n / res["tau"].max(), n / res["tau"].max() / res["wall"]

ess_A, essps_A = ess_per_sec(res_A)
ess_B, essps_B = ess_per_sec(res_B)
ess_C = result.diagnostics["ess"]
rows = [("A  TransitModel + emcee (7-D)", res_A["evals"], res_A["wall"], ess_A, essps_A),
        ("B  Marginal + emcee (5-D)", res_B["evals"], res_B["wall"], ess_B, essps_B),
        ("C  Marginal + fit() (IS)", result.diagnostics["n_eval"], wall_C, ess_C, ess_C / wall_C)]
print(f"{'':32s} {'evaluations':>12s} {'wall [s]':>9s} {'ESS':>8s} {'ESS/s':>9s}")
for r in rows:
    print(f"{r[0]:32s} {r[1]:12d} {r[2]:9.1f} {r[3]:8.0f} {r[4]:9.1f}")
print(f"\nevaluations to convergence, A / B: {res_A['evals'] / res_B['evals']:.1f}x;   A / C: {res_A['evals'] / result.diagnostics['n_eval']:.0f}x")
print(f"ESS per second, B / A: {essps_B / essps_A:.1f}x;   C / A: {(ess_C / wall_C) / essps_A:.0f}x")
````

## Same budget (fixed number of steps), including batman

Notebook 02 runs batman + emcee and pizzetti + emcee for a fixed number of steps with identical likelihoods
(default settings of each package). Here the same is done for batman (**A₀**), pizzetti (**A**) and the
marginalised model (**B**), alternating, best of three. Note that 3000 steps is short of emcee's 50τ
recommendation for A (and τ is underestimated in short runs); the convergence-based comparison above is
the fair measure of effort, this one shows the cost per step.

````python
def make_log_prob(module):
    # example 01's likelihood, for either package (default model settings)
    prm = params(module, **truth)
    m = module.TransitModel(prm, t)
    def lp(th):
        t0, per, rp, a, b, q1, q2 = th
        if not (0 < rp < 0.5 and 1.5 < a < 100 and 0 <= b < 1 + rp and 0 < q1 < 1 and 0 < q2 < 1):
            return -np.inf
        prm.t0, prm.per, prm.rp, prm.a = t0, per, rp, a
        prm.inc = np.degrees(np.arccos(b / a))
        prm.u = list(to_u(q1, q2))
        r = y - m.light_curve(prm)
        return -0.5 * np.dot(r, r) / sigma**2
    lp(p0[0])                                         # warm-up
    return lp

configs = {"A0 batman + emcee (7-D)": lambda lp=make_log_prob(batman): emcee.EnsembleSampler(nwalkers, 7, lp),
           "A  pizzetti + emcee (7-D)": lambda lp=make_log_prob(pizzetti): emcee.EnsembleSampler(nwalkers, 7, lp),
           "B  Marginal + emcee (5-D)": lambda: model.EnsembleSampler(nwalkers)}
walls = {k: [] for k in configs}
for rep in range(1 if SMOKE else 3):
    for k in (list(configs) if rep % 2 == 0 else list(configs)[::-1]):
        s = configs[k]()
        s.random_state = np.random.RandomState(2024).get_state()
        t1 = time.perf_counter()
        s.run_mcmc(p0, 500 if SMOKE else 3000, progress=False)
        walls[k].append(time.perf_counter() - t1)
        if rep == 0:
            ch = s.get_chain(expand=False) if k.startswith("B") else s.get_chain()
            walls[k + " tau"] = emcee.autocorr.integrated_time(ch, quiet=True, tol=0).max()
print(f"{'3000 steps x 32 walkers':28s} {'best wall [s]':>13s} {'us/eval':>8s} {'max tau':>8s} {'ESS/s':>7s}")
for k in configs:
    w, tau = min(walls[k]), walls[k + " tau"]
    print(f"{k:28s} {w:13.2f} {w / 96000 * 1e6:8.1f} {tau:8.0f} {64000 / tau / w:7.0f}")
wb, wp, wm = (min(walls[k]) for k in configs)
print(f"\nspeed-up per step vs batman: pizzetti {wb / wp:.1f}x, marginal {wb / wm:.1f}x")
````

## Posteriors

A fixes the baseline flux to 1 (as in notebook 02); B and C fit it (it is integrated out). The geometry and
limb-darkening posteriors agree within their Monte Carlo noise.

````python
sel_A = chain_A
lab = ["rp", "a", "b", "q1", "q2"]
X_A = chain_A[:, [names.index(k) for k in lab]]
X_B = chain_B[:, [sampler_B.labels.index(k) for k in lab]]
X_C = result.samples[:, [result.labels.index(k) for k in lab]]          # equal weight (for the table)
XW_C = result.weighted_samples[:, [result.labels.index(k) for k in lab]]  # exact weighted draws (plot)

print(f"{'':4s} {'truth':>8s} {'A (median ± sd)':>22s} {'B':>22s} {'C':>22s} {'(B-A)/σ':>8s} {'(C-A)/σ':>8s}")
tq = {"rp": truth["rp"], "a": truth["a"], "b": truth["b"], "q1": th0[5], "q2": th0[6]}
for j, k in enumerate(lab):
    a, b, c = X_A[:, j], X_B[:, j], X_C[:, j]
    print(f"{k:4s} {tq[k]:8.4f} {np.median(a):10.5f} ± {a.std():.5f} {np.median(b):10.5f} ± {b.std():.5f} "
          f"{np.median(c):10.5f} ± {c.std():.5f} {(np.median(b) - np.median(a)) / a.std():+8.2f} {(np.median(c) - np.median(a)) / a.std():+8.2f}")
````

## FitResult: what `fit()` returns

`result.samples` are equal-weight draws with column names `result.labels`: the sampled geometry, the
limb-darkening coefficients (`q1, q2` by default; `fit(ld="u")` gives `u1, u2`) and the baseline `f0`.
`result.weighted_samples` with `result.weights` are the exact weighted draws, best for histograms;
`result.summary()` and `result.corner()` give a quick look; `result.diagnostics` says how the draws were made.

````python
print(result.labels)
print({k: (round(v, 3) if isinstance(v, float) else v) for k, v in result.diagnostics.items() if k != "tau"})
res_u = model.fit(seed=2, n=2000, ld="u")
print(res_u.summary())
````

## Drawing the coefficients for your own geometry samples: `expand`

`expand(X)` takes rows of the sampled parameters (here from any source: a grid, an optimiser, another sampler) and
returns them with the limb-darkening coefficients and the baseline drawn from their exact conditional posterior.

````python
X = np.array([[0.0, 3.5, 0.1, 9.0, 0.35]] * 3)          # t0, per, rp, a, b
rows, labels, w = model.expand(X, ld="u", rng=np.random.default_rng(0))
print(labels); print(np.round(rows, 4))
````

## Priors

- **Geometry**: `bounds={"rp": (0, 0.5), ...}` sets uniform box priors; `log_prior=f` adds any vectorised log prior
  `f(dict of arrays) -> array` on the physical parameters.
- **Limb darkening**: `ld_prior="physical"` (default) is flat over the law's physical region (positive intensity, darkening
  towards the limb); `"flat"` drops that constraint; a dict `{"mean": (u1, u2), "cov": ...}` is a Gaussian prior (quadratic law,
  exact); any callable `f(u1, u2) -> log prior` is applied by importance reweighting.

````python
def fit_with(**kw):
    m = pizzetti.MarginalTransitModel(p_true, t, y, sigma, free=["t0", "per", "rp", "a", "b"],
                                      bounds={"rp": (0, 0.5), "a": (1.5, 100)}, **kw)
    r = m.fit(seed=3, n=2000, ld="u")
    j = [r.labels.index(k) for k in ("rp", "b", "u1", "u2")]
    return np.median(r.samples[:, j], axis=0), r.samples[:, j].std(axis=0)

priors = {
    "physical (default)": {},
    "flat": dict(ld_prior="flat"),
    "Gaussian u = (0.40, 0.25) ± 0.05": dict(ld_prior={"mean": (0.40, 0.25), "cov": np.diag([0.05**2, 0.05**2])}),
    "callable: u2 near 0.25": dict(ld_prior=lambda u1, u2: -0.5 * ((u2 - 0.25) / 0.05)**2),
    "log_prior: a near 9.0 ± 0.1": dict(log_prior=lambda P: -0.5 * ((P["a"] - 9.0) / 0.1)**2),
}
print(f"{'prior':36s} {'rp':>16s} {'b':>14s} {'u1':>14s} {'u2':>14s}")
for name, kw in priors.items():
    med, sd = fit_with(**kw)
    print(f"{name:36s}" + "".join(f"{m:9.4f} ± {s:.3f}" for m, s in zip(med, sd)))
````

A prior on the stellar density (here, on a/R⋆) or on the limb darkening tightens the impact parameter, which is
degenerate with both.

## Power-2 with free α

For `limb_dark="power2"`, `params.u[1]` is the exponent α, fixed unless `"alpha"` is in `free`; then it is sampled
with the geometry.

````python
p2 = pizzetti.TransitParams()
p2.t0, p2.per, p2.rp, p2.a, p2.inc, p2.ecc, p2.w = p_true.t0, p_true.per, p_true.rp, p_true.a, p_true.inc, 0.0, 90.0
p2.limb_dark, p2.u = "power2", [0.6, 0.6]
m2 = pizzetti.MarginalTransitModel(p2, t, y, sigma, free=["t0", "per", "rp", "a", "b", "alpha"],
                                   bounds={"rp": (0, 0.5), "a": (1.5, 100), "alpha": (0.1, 3.0)})
r2 = m2.fit(seed=4, n=4000, ld="u")
print(r2.diagnostics["method"], f"khat = {r2.diagnostics['khat']:.2f}")
print(r2.summary())
````

## Secondary eclipses

With `transittype="secondary"` the linear parameters are the planet's flux `fp` (constrained to fp ≥ 0) and the baseline;
the timing parameter is `t_secondary`.

````python
ps = pizzetti.TransitParams()
ps.t0, ps.per, ps.rp, ps.a, ps.inc, ps.ecc, ps.w = 0.0, 3.5, 0.1, 9.0, p_true.inc, 0.0, 90.0
ps.limb_dark, ps.u, ps.fp, ps.t_secondary = "uniform", [], 400e-6, 1.75
te = (1.75 + np.arange(5)[:, None] * 3.5 + np.arange(-0.13, 0.13, 2.0 / 1440)[None, :]).ravel()
ye = pizzetti.TransitModel(ps, te, transittype="secondary").light_curve(ps)
ye = ye + 200e-6 * np.random.default_rng(5).standard_normal(te.size)
me = pizzetti.MarginalTransitModel(ps, te, ye, 200e-6, free=["t_secondary"], transittype="secondary")
re = me.fit(seed=5, method="emcee", n=2000)
print(re.summary())
````

## What changed in the user code (A → B)

```python
# A (notebook 03)                                   # B (new)
model = pizzetti.TransitModel(prm, t)               model = pizzetti.MarginalTransitModel(p_true, t, y, sigma,
def log_prob(th): ...  # ~10 lines                          free=["t0", "per", "rp", "a", "b"],

```

Or, for the fastest route, `result = model.fit()`.

**Scope:** this notebook uses the quadratic law, but `MarginalTransitModel` supports every law pizzetti knows except the
exponential law (uniform, linear, quadratic, square-root, logarithmic, power-2 with fixed or free α, nonlinear), for primary
transits (rp > 0, circular or eccentric orbits) and secondary eclipses, with supersampling, per-point errors, box priors,
a vectorised `log_prior` on the geometry, and a Gaussian prior on (u1, u2) (exact; quadratic law only, other laws take a
callable `ld_prior`). Inverse transits (rp < 0) and the exponential law raise `NotImplementedError`.
[Notebook 05](05_choosing_a_limb_darkening_law.ipynb) fits the same data with the other laws.

## What you learned

- `MarginalTransitModel(params, t, y, yerr, free=[...])` integrates the limb darkening and the baseline out; `fit()` returns a `FitResult` in about a second.
- `EnsembleSampler` is an emcee sampler on the geometry only; `expand` draws the coefficients for any geometry samples.
- Priors: `bounds`, `log_prior` (geometry), `ld_prior` (physical, flat, Gaussian, callable).
- Every law, power-2 with free α, and secondary eclipses work the same way.
